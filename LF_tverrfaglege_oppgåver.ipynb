# Løysningsforslag tverrfaglege oppgåver

### Oppgåve 1

```python
plastavfall = 431 # kg for eksempel
prosent = 10 
vekstfaktor = 1 - prosent / 100 # vekstfaktor for 10 prosent nedgong blir rekna ut til 0.9
år = 2025 # me startar i året 2025

while plastavfall > 100: # så lenge plastavfallet er over 100 (kg), så skal ...
    plastavfall = plastavfall*vekstfaktor # plastavfallet minkes med 10 %
    år = år + 1 # for kvart år som går

# b)
print('Simen når målet sitt med å redusere plastavfall til 100 (kg for eksempel) i', år)
```

### Oppgåve 2

```python
dager = 0 # startar på dag 0
kaniner = 4 # det er 4 kaniner
vekstfaktor = 1.1 # vekstfaktoren er 1.1, altså 10 % auke
gulrøtter = 3000 # det finst 3000 gulrøtter

while gulrøtter > 0: # så lenge det er gulrøtter tilgjengelig, så skal...
    kaniner = kaniner*vekstfaktor # kaninene vokse med 10 % (for kvar dag)
    gulrøtter = gulrøtter - kaniner*4 # kvar dag et kaninene 4 gulrøtter kvar
    dager = dager + 1 # dagene går

print(kaniner) #skriv ut kor mange kaniner det er i populasjonen
print(dager) # skriv ut hvor mange dager gulrøttene holdt
```

### Oppgåve 3

```python
# Dette er eit program som brukar både input kommandoen og lister

m = float(input('Hvor mange mg milgøgift er det i krill?')) 
# ber om mg miljøgift i krill og lagrer det som ein float, altså eit desimaltal, i variabelen m

næringskjede = ['krill', 'lodde', 'sild', 'måke']
# lager ei liste med alle dyra i næringskjeda. Dei er tekst, altså strings, markert med anførselsetegn ''

for ledd in range(len(næringskjede)): # for kvart ledd i næringskjeda, så skal...
    # kommandoen len() står for length og finner dermed lengda av lista
    print('Det er', m, 'mg miljøgift i', næringskjede[ledd]) # svaret skrives ut
    m = m*10 # konsentrasjonen av miljøgift ganges med 10 for kvart ledd i næroingskjeda
    
```

### Oppgåve 4

```python
matsvinn = 160 # kg mat i året som familien kastar
mål = matsvinn / 2 # målet er å halvere det årlege matsvinnet
vf = 0.87 # Sofie reknar med at familien skal klare å redusere svinnet med 13 % pr. år

år = 2025 # me startar i 2025

while matsvinn > mål: # så lenge målet ikkje er nådd, så skal...
    matsvinn = matsvinn*vf # matsvinnet reduseres med 13 %
    år = år + 1 # for kvart år

print(år) # skriv ut året målet er nådd
print(matsvinn) # skriv ut kg matsvinn familien har det året. Me ser det er rett under 80 kg,
                # som var målet
```
