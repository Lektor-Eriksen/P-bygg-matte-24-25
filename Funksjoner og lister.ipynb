## Døme 1

Preben syklar med ein konstant fart på 16 km/t. 

a) Lag ein funksjon s(t) som reknar ut kor langt han har sykla. 

b) Kor langt har han sykla etter 2.5 time?

c) Kor lenge må han sykle for å kome seg rundt Stord-øyo som er ca. 110 km? 
   Tips: Det kan vere lurt å bruke ei while-løkke


```python
# a) 

def s(t):
    return 16*t
```

```python
# b) 

s(2.5)
```

```python
# c) 

tid = 0

while s(tid) < 110:
    tid = tid + 0.001

print('etter', round(tid,2), 'timar så har Preben sykla', round(s(tid),2), 'km')
```
