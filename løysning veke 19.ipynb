### oppgåve 2

```python
beløp = 5000 # startbeløp
vekstfaktor = 1.02 # renta er på 2 %
år = 0 # startar sparingen

while beløp < 100000: # skal spare til 100 000 kr
    år = år + 1 # for kvart år
    beløp = beløp*vekstfaktor # me får renter på kontoen
    beløp = beløp + 10000 # frå og med neste år sparar 10 000 kr kvart år

print(round(beløp, 0)) # runder av beløpet til heile kroner
print(år) # skriv ut kor mange år det tok

# b)
# dei fortel kor lang tid det tek å få meir enn 100 000 kr på konto,
# og nøyaktig kor mykje det er på kontoen
```

### oppgåve 3


```python
a = 1 # variabelen a blir satt til å vere 1
while a < 10: # mens a er mindre enn 10, så skal....
    svar = a**(1/2) # svaret er a^(1/2) altså så tek me eigentleg kvadratrota
    a = a + 1 # me går til neste heile tal
    print(svar) # b) me skriv ut kvadratrota til dei første 9 tala
    
```
