```python
a = 3

b = 4

if a < b: # viss a er mindre enn b, så
    print(a, 'er mindre enn', b) 
    # skriv ut dette
elif a > b: # viss a er større enn b, så
    print(a, 'er større enn', b)
    # skriv ut dette
else: # elles, så
    print(a, 'er lik', b)
    # skriv ut dette
```
